```python
# 1. Digital Image Basics
# What is a Digital Image?

# A digital image is simply a matrix (grid) of tiny squares called pixels.

# Each pixel stores numerical values that represent color or brightness.

# For example, a small grayscale image might look like this:

# 120	130	150	180
# 125	135	155	185
# 128	140	160	190
# 130	145	170	200

# Here:

# Every number is the intensity of one pixel.
# Lower values are darker.
# Higher values are brighter.


```

```python
# What is a Pixel?
# A pixel (Picture Element) is the smallest unit of a digital image.

# Imagine zooming into a photograph until you can see tiny squares. Each square is one pixel.

# Each pixel contains information.

# For a grayscale image, each pixel stores one value.



# 0      = Black
# 50     = Dark Gray
# 128    = Gray
# 200    = Light Gray
# 255    = White
```

```python
# Color Images

# Color images are slightly different.

# Instead of storing one value, each pixel stores three values.



# [255,0,0] -> Red
# [0,255,0]->Green
# [0,0,255]-> Blue


# A color image is therefore a 3D array.

# Height × Width × Channels


# Example

# 2 × 2 image

# [
#  [[255,0,0], [0,255,0]],

#  [[0,0,255], [255,255,0]]
# ]
```

```python
# Image Dimensions

# Suppose an image is

# 1920 × 1080

# 1920 columns
# 1080 rows

# Total pixels

# 1920 × 1080

# = 2,073,600 pixels

# ≈ 2.07 Million Pixels
# This is called a 2 Megapixel image.
```
