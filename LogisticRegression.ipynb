```python
# 1️⃣ Binomial (Binary) Logistic Regression
# Concept

# Used when the target has only 2 classes

# Output is a probability between 0 and 1

# Uses Sigmoid function

# Examples

# Pass / Fail

# Yes / No

# Spam / Not Spam

# Disease / No Disease

# Example Dataset
# Study Hours	    Passed
# 2	                   0
# 5	                   1
# 1	                   0
# 6	                   1

```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Features (X) and Target (y)
X = np.array([[2], [5], [1], [6], [3], [4]])
y = np.array([0, 1, 0, 1, 0, 1])

# Model
model = LogisticRegression()
model.fit(X, y)

# Prediction
hours = [[6]]
prediction = model.predict(hours)
probability = model.predict_proba(hours)

print("Prediction (0=Fail, 1=Pass):", prediction)
print("Probability:", probability)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Training data
X = np.array([[2], [5], [1], [6], [3], [4],[0],[7]])
y = np.array([0, 1, 0, 1, 0, 1,0,1])

# Train model
model = LogisticRegression()
model.fit(X, y)

# Create smooth X values for curve
X_curve = np.linspace(0, 7, 200).reshape(-1, 1)

# Predict probabilities
y_prob = model.predict_proba(X_curve)[:, 1]

# Plot
plt.figure()
plt.plot(X_curve, y_prob)
plt.scatter(X, y)
plt.axhline(0.5)
plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Sigmoid (S-Curve) - Logistic Regression")
plt.show()
```

```python
# 2️⃣ Multinomial Logistic Regression

# Concept

# Used when the target has more than 2 classes

# Classes are unordered

# Uses Softmax function

# Extension of binary logistic regression

# Examples

# Fruit type → Apple / Banana / Orange

# Digit recognition → 0–9

# Example Dataset
# Marks	   Grade
# 85	      A
# 65	      B
# 45	      C
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Features
X = np.array([[90], [70], [50], [85], [60], [40]])

# Labels: A=0, B=1, C=2
y = np.array([0, 1, 2, 0, 1, 2])

# Multinomial model
model = LogisticRegression()
model.fit(X, y)

# Prediction
marks = [[100]]
prediction = model.predict(marks)
probability = model.predict_proba(marks)

print("Predicted Class:", prediction)
print("Class Probabilities:", probability)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Training data
X = np.array([[40], [50], [60], [70], [80], [90]])
y = np.array([2, 2, 1, 1, 0, 0])  # A=0, B=1, C=2

# Train multinomial model
model = LogisticRegression(solver='lbfgs')
model.fit(X, y)

# Smooth range for plotting
X_curve = np.linspace(35, 95, 300).reshape(-1, 1)

# Predict probabilities (Softmax output)
probs = model.predict_proba(X_curve)

# Plot probability curves
plt.figure()
for i in range(probs.shape[1]):
    plt.plot(X_curve, probs[:, i])

plt.xlabel("Marks")
plt.ylabel("Probability")
plt.title("Multinomial Logistic Regression (Softmax Curves)")
plt.show()
```

```python
# 3️⃣ Ordinal Logistic Regression
# Concept

# Used when the target has ordered categories

# Order matters

# Standard LogisticRegression ❌ does NOT handle order

# Uses Ordinal Regression (Proportional Odds Model)

# Examples

# Ratings → Poor < Average < Good < Excellent

# Satisfaction → Low < Medium < High

# Education → School < Graduate < Post-Graduate
```

```python
import numpy as np
import mord

# Features
X = np.array([[2], [4], [6], [8], [3], [5], [7]])

# Labels (ordered)
# Poor=0, Average=1, Good=2, Excellent=3
y = np.array([0, 1, 2, 3, 1, 2, 3])

# Ordinal Regression Model
model = mord.LogisticIT()
model.fit(X, y)

# Prediction
hours = np.array([[8]])
prediction = model.predict(hours)

print("Predicted Performance Level:", prediction)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import mord

# Training data
X = np.array([[2], [4], [6], [8], [3], [5], [7]])
y = np.array([0, 1, 2, 3, 1, 2, 3])  # Poor=0, Average=1, Good=2, Excellent=3

# Train ordinal model
model = mord.LogisticIT()
model.fit(X, y)

# Smooth X range
X_curve = np.linspace(1, 9, 300).reshape(-1, 1)

# Predict probabilities
probs = model.predict_proba(X_curve)

# Plot probability curves
plt.figure()
for i in range(probs.shape[1]):
    plt.plot(X_curve, probs[:, i])

plt.xlabel("Study Hours")
plt.ylabel("Probability")
plt.title("Ordinal Logistic Regression (Ordered Probability Curves)")
plt.show()
```
